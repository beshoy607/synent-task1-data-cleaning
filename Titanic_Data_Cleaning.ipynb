# Synent Technologies – Task 1: Data Cleaning & Preprocessing

**Internship Domain:** Data Science  
**Task:** Data Cleaning & Preprocessing  
**Dataset:** Titanic Dataset

## Objective
Clean and prepare the Titanic dataset for analysis by:
- Handling missing values
- Removing duplicate rows
- Converting data types
- Renaming columns

The final output is a clean dataset ready for analysis.

## 1. Import Libraries

```python
import pandas as pd
pd.set_option("display.max_columns",None)
```

## 2. Load the Dataset

```python
df= pd.read_csv("Titanic-Dataset.csv")
print("Dataset shape:",df.shape)
df.head()
```

## 3. Inspect the Dataset

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:",df.duplicated().sum())
```

## 4. Handle Missing Values

The missing values are handled as follows:

- **Age:** missing values are replaced with the median age because it is less affected by extreme values.
- **Embarked:** missing values are replaced with the most frequent value (mode).
- **Cabin:** missing values are replaced with `"Unknown"` because many cabin values are unavailable, so dropping the column would lose useful information.

```python
print("Missing values before cleaning:")
print(df.isnull().sum())
df["Age"]= df["Age"].fillna(df["Age"].median())
df["Embarked"] =df["Embarked"].fillna(df["Embarked"].mode()[0])
df["Cabin"]= df["Cabin"].fillna("Unknown")

print("\nMissing values after handling:")
print(df.isnull().sum())
```

## 5. Remove Duplicate Rows

```python
duplicates_before= df.duplicated().sum()
print("Duplicate rows before removal:",duplicates_before)
df= df.drop_duplicates().copy()
print("Duplicate rows after removal:",df.duplicated().sum())
```

## 6. Convert Data Types

Numeric columns are explicitly converted to appropriate numeric types. This makes the dataset consistent and ready for later analysis.

```python
df["PassengerId"]= df["PassengerId"].astype("int64")
df["Survived"]= df["Survived"].astype("int64")
df["Pclass"]= df["Pclass"].astype("int64")
df["SibSp"]= df["SibSp"].astype("int64")
df["Parch"]= df["Parch"].astype("int64")
df["Fare"]= df["Fare"].astype("float64")
print(df.dtypes)
```

## 7. Rename Columns

```python
df = df.rename(columns={
    "PassengerId":"passenger_id",
    "Survived":"survived",
    "Pclass":"passenger_class",
    "Name":"name",
    "Sex":"sex",
    "Age":"age",
    "SibSp":"siblings_spouses",
    "Parch":"parents_children",
    "Ticket":"ticket",
    "Fare":"fare",
    "Cabin":"cabin",
    "Embarked":"embarked"
})

print("New column names:")
print(df.columns.tolist())
```

## 8. Verify the Cleaned Dataset

```python
print("Final dataset shape:",df.shape)
print("Total missing values:",df.isnull().sum().sum())
print("Total duplicate rows:",df.duplicated().sum())
df.head()
```

## 9. Save the Cleaned Dataset

The cleaned dataset is saved as `titanic_cleaned.csv` and is ready for further analysis.

```python
df.to_csv("titanic_cleaned.csv",index=False)
print("Cleaned dataset saved successfully as titanic_cleaned.csv")
```

## 10. Final Summary

After preprocessing:

- All missing values were handled.
- Duplicate rows were removed.
- Data types were reviewed and converted where needed.
- Column names were renamed using a consistent format.
- The final dataset contains **891 rows and 12 columns**.
- The final dataset contains **0 missing values** and **0 duplicate rows**.

The cleaned dataset is ready for analysis.
